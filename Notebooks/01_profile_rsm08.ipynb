## Calculate our own retail growth metrics

```python
retail_clean["mom_growth_pct"] = (
    retail_clean["retail_sales_index"]
    .pct_change() * 100
)

retail_clean.head(10) #this calculates the (current month - previous month) / previous month * 100 to get the percentage change from the previous month to the current month.
```

```python
retail_clean["yoy_growth_pct"] = (
    retail_clean["retail_sales_index"]
    .pct_change(periods=12) * 100
)

retail_clean.head(15)
```

## CSO-vs-our-calculation validation step

```python
growth_series = df[
    (df["NACE Group"] == "All retail businesses") &
    (df["UNIT"] == "%")
].copy()

growth_series[
    ["STATISTIC", "Statistic Label"]
].drop_duplicates()
```

```python
official_growth = df[
    (df["NACE Group"] == "All retail businesses") &
    (df["UNIT"] == "%") &
    (df["STATISTIC"].isin(["RSM08C05", "RSM08C06"]))
].copy()

official_growth = official_growth[
    ["STATISTIC", "Month", "VALUE"]
]

official_growth.head(15)
```

```python
official_growth.shape
```

```python
official_growth.head(15)
```

```python
official_growth.shape
```

```python
official_growth_wide = official_growth.pivot(
    index="Month",
    columns="STATISTIC",
    values="VALUE"
).reset_index()

official_growth_wide
```

```python
official_growth_wide = official_growth_wide.rename(
    columns={
        "RSM08C05": "cso_mom_growth_pct",
        "RSM08C06": "cso_yoy_growth_pct"
    }
)

official_growth_wide.head(15)
```

```python
official_growth_wide["date"] = pd.to_datetime(
    official_growth_wide["Month"],
    format="%Y %B"
)

official_growth_wide = official_growth_wide.drop(
    columns=["Month"]
)

official_growth_wide = official_growth_wide.sort_values("date")

official_growth_wide.reset_index(drop=True, inplace=True)

official_growth_wide.head(15)
```

merge it with our calculation

```python
validation = retail_clean.merge(
    official_growth_wide,
    on="date",
    how="left"
)

validation.head(15)
```

```python
validation["mom_difference"] = (
    validation["mom_growth_pct"]
    - validation["cso_mom_growth_pct"]
)

validation["yoy_difference"] = (
    validation["yoy_growth_pct"]
    - validation["cso_yoy_growth_pct"]
)

validation.head(15)
```

check the maximum difference

```python
validation["mom_difference"].abs().max()
```

```python
validation["yoy_difference"].abs().max()
```

check the average difference


```python
validation["mom_difference"].abs().mean()
```

```python
validation["yoy_difference"].abs().mean()
```

final validation test , make our comparison cleaner by rounding our calculations to one decimal point

```python
validation["mom_match"] = (
    validation["mom_growth_pct"].round(1)
    == validation["cso_mom_growth_pct"]
)

validation["yoy_match"] = (
    validation["yoy_growth_pct"].round(1)
    == validation["cso_yoy_growth_pct"]
)
```

```python
validation["mom_match"].value_counts(dropna=False)
```

```python
validation["yoy_match"].value_counts(dropna=False)
```

```python
mom_mismatches = validation[
    validation["mom_match"] == False
][
    [
        "date",
        "retail_sales_index",
        "mom_growth_pct",
        "cso_mom_growth_pct",
        "mom_difference"
    ]
]

mom_mismatches
```

```python
yoy_mismatches = validation[
    validation["yoy_match"] == False
][
    [
        "date",
        "retail_sales_index",
        "yoy_growth_pct",
        "cso_yoy_growth_pct",
        "yoy_difference"
    ]
]

yoy_mismatches
```

```python
validation_valid = validation[
    validation["cso_mom_growth_pct"].notna()
].copy()

validation_valid["mom_difference"] = (
    validation_valid["mom_growth_pct"]
    - validation_valid["cso_mom_growth_pct"]
)

validation_yoy_valid = validation[
    validation["cso_yoy_growth_pct"].notna()
].copy()

validation_yoy_valid["yoy_difference"] = (
    validation_yoy_valid["yoy_growth_pct"]
    - validation_yoy_valid["cso_yoy_growth_pct"]
)
```

```python
print(
    "Maximum MoM difference:",
    validation_valid["mom_difference"].abs().max()
)

print(
    "Maximum YoY difference:",
    validation_yoy_valid["yoy_difference"].abs().max()
)
```

```python
validation_valid["mom_within_tolerance"] = (
    validation_valid["mom_difference"].abs() <= 0.11
)

validation_yoy_valid["yoy_within_tolerance"] = (
    validation_yoy_valid["yoy_difference"].abs() <= 0.11
)
```

```python
print(
    validation_valid["mom_within_tolerance"].value_counts()
)

print(
    validation_yoy_valid["yoy_within_tolerance"].value_counts()
)
```

```python
validation_valid[
    validation_valid["mom_within_tolerance"] == False
][
    [
        "date",
        "retail_sales_index",
        "mom_growth_pct",
        "cso_mom_growth_pct",
        "mom_difference"
    ]
]
```

```python
validation_valid[
    validation_valid["mom_difference"].between(0.11, 0.12)
][
    [
        "date",
        "mom_growth_pct",
        "cso_mom_growth_pct",
        "mom_difference"
    ]
]
```
