```python
import pandas as pd

df = pd.read_csv(
    "../data/raw/mum01_raw.csv",
    encoding="utf-8-sig"
)

df.columns = df.columns.str.replace(
    'ï»¿"',
    '',
    regex=False
)

df.columns = df.columns.str.replace(
    '"',
    '',
    regex=False
)

print("Rows:", len(df))
print("Columns:", len(df.columns))
print("\nColumn names:")
print(df.columns.tolist())
```

```python
print(df.head(10).to_string())
```

## so we use Seasonally Adjusted Monthly Unemployment — age 15–74 — Both sexes — Thousand


# That gives us the overall number of unemployed people, seasonally adjusted, without splitting the population by sex.

```python
unemployment = df[
    (df["STATISTIC"] == "MUM01C01") &
    (df["Age Group"] == "15 - 74 years") &
    (df["Sex"] == "Both sexes")
].copy()

print(unemployment.shape)
print(unemployment.head(10).to_string())
print(unemployment.tail(10).to_string())
```

```python
print(unemployment[["Month", "UNIT", "VALUE"]].to_string(index=False))
```

# check for the missing values and duplicates

```python
print("Missing values:")
print(unemployment[["Month", "VALUE"]].isna().sum())

print("\nDuplicate months:")
print(unemployment["Month"].duplicated().sum())

print("\nDate range:")
print(unemployment["Month"].min(), "to", unemployment["Month"].max())
```

```python
print(
    df[df["Month"].str.contains("2026")][
        ["STATISTIC", "Statistic Label", "Month",
         "Age Group", "Sex", "UNIT", "VALUE"]
    ].tail(50).to_string(index=False)
)
```

```python
print("\nLatest month in entire raw dataset:")
print(df["Month"].max())

print("\nLatest 20 unique months:")
print(df["Month"].drop_duplicates().tail(20).tolist())
```

```python
unemployment = df[
    (df["STATISTIC"] == "MUM01C02") &
    (df["Age Group"] == "15 - 74 years") &
    (df["Sex"] == "Both sexes")
].copy()

print(unemployment.shape)
print(unemployment.head())
print(unemployment.tail())
```

```python
print(
    unemployment[["Month", "UNIT", "VALUE"]]
    .tail(15)
    .to_string(index=False)
)
```

```python
print("Missing values:")
print(unemployment[["Month", "VALUE"]].isna().sum())

print("\nDuplicate months:")
print(unemployment["Month"].duplicated().sum())

print("\nDate range:")
print(unemployment["Month"].min(), "to", unemployment["Month"].max())
```

```python
unemployment = unemployment[["Month", "VALUE"]].copy()

unemployment = unemployment.rename(columns={
    "Month": "date",
    "VALUE": "unemployment_rate"
})

unemployment["date"] = pd.to_datetime(
    unemployment["date"],
    format="%Y %B"
)

unemployment = unemployment.sort_values("date").reset_index(drop=True)

print(unemployment.head())
print(unemployment.tail())
print("Shape:", unemployment.shape)
```

```python
print("Missing values:")
print(unemployment.isna().sum())

print("\nDuplicate dates:")
print(unemployment["date"].duplicated().sum())

print("\nDate range:")
print(unemployment["date"].min(), "to", unemployment["date"].max())

print("\nUnemployment rate range:")
print(
    unemployment["unemployment_rate"].min(),
    "to",
    unemployment["unemployment_rate"].max()
)
```

```python
unemployment.to_csv(
    "../data/processed/unemployment.csv",
    index=False
)

print("Saved to data/processed/unemployment.csv")
```
