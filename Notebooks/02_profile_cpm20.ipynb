# isolate annual headline inflation

## we want CPM20C08 + all items  That gives us the 12-month percentage change in the Consumer Price Index for All Items.

```python
cpi = df[
    (df["STATISTIC"] == "CPM20C08") &
    (df["Commodity Group"] == "All Items")
].copy()
```

```python
print(cpi.head(10).to_string())
```

```python
print(cpi.tail(10).to_string())
```

```python
print(cpi.shape)
```

```python
print(cpi[["Month", "VALUE", "UNIT"]].to_string(index=False))
```

```python
inflation = cpi[
    ["Month", "VALUE"]
].copy()

inflation = inflation.rename(columns={
    "Month": "date",
    "VALUE": "inflation_rate"
})

inflation["date"] = pd.to_datetime(
    inflation["date"],
    format="%Y %B"
)

inflation = inflation.sort_values("date").reset_index(drop=True)
```

```python
print(inflation.head())
print(inflation.tail())
print(inflation.dtypes)
```

```python
print("Rows:", len(inflation))
print("Missing values:")
print(inflation.isna().sum())

print("\nDuplicate dates:", inflation["date"].duplicated().sum())

print("\nDate range:")
print(inflation["date"].min(), "to", inflation["date"].max())

print("\nInflation range:")
print(inflation["inflation_rate"].min(), "to", inflation["inflation_rate"].max())
```

```python
print("\nDuplicate dates:")
print(
    inflation[inflation["date"].duplicated(keep=False)]
)
```

```python
import os

os.makedirs("../data/processed", exist_ok=True)

inflation.to_csv(
    "../data/processed/inflation.csv",
    index=False
)

print("Saved successfully.")
```
