One important point: the master data is split by Banks and presumably Non-Banks, so we should first determine exactly how the current Central Bank Table 1 - PDH totals are constructed. We don't want to accidentally use only bank mortgages.

# Step 1 — Inspect the available entity groups

```python
master1.columns = master1.iloc[0]
master1 = master1.iloc[1:].reset_index(drop=True)

print(master1.columns)
print(master1.head())
```

```python
print(master1["Entity Group"].unique())
print(master1["Format"].unique())
print(master1["Category"].unique())
```

we'll focus on PDH (Principal Dwelling Houses) and combine Banks + Non-Banks.

# Step 1 — Inspect the PDH arrears rows

```python
arrears_rows = master1[
    (master1["Format"] == "PDH") &
    (master1["Category"] == "Arrears")
]

print(
    arrears_rows[
        ["Entity Group", "Row Position", "Row", "Column"]
    ].drop_duplicates().to_string(index=False)
)
```

```python
import pandas as pd

# Make sure Amt is numeric
master1["Amt"] = pd.to_numeric(master1["Amt"], errors="coerce")

# Make sure Reporting Date is a proper date
master1["Reporting Date"] = pd.to_datetime(
    master1["Reporting Date"],
    errors="coerce"
)

# ---------------------------------------------------------
# 1. Total PDH mortgage accounts outstanding
# ---------------------------------------------------------

outstanding = master1[
    (master1["Format"] == "PDH") &
    (master1["Category"] == "Total Outstanding") &
    (master1["Column"] == "Number") &
    (master1["Entity Group"].isin(["Banks", "Non-Banks"]))
].copy()

outstanding = (
    outstanding
    .groupby("Reporting Date")["Amt"]
    .sum()
    .reset_index()
    .rename(columns={
        "Reporting Date": "date",
        "Amt": "pdh_mortgage_accounts"
    })
)


# ---------------------------------------------------------
# 2. Total PDH mortgage accounts in arrears
# ---------------------------------------------------------

total_arrears = master1[
    (master1["Format"] == "PDH") &
    (master1["Category"] == "Arrears") &
    (master1["Row Position"] == 2) &
    (master1["Column"] == "Number") &
    (master1["Entity Group"].isin(["Banks", "Non-Banks"]))
].copy()

total_arrears = (
    total_arrears
    .groupby("Reporting Date")["Amt"]
    .sum()
    .reset_index()
    .rename(columns={
        "Reporting Date": "date",
        "Amt": "pdh_arrears_accounts"
    })
)


# ---------------------------------------------------------
# 3. PDH mortgage accounts in arrears >90 days
# ---------------------------------------------------------

# These are the NON-OVERLAPPING duration categories.
arrears_90_plus_positions = [4, 6, 7, 9, 10, 11]

arrears_90_plus = master1[
    (master1["Format"] == "PDH") &
    (master1["Category"] == "Arrears") &
    (master1["Row Position"].isin(arrears_90_plus_positions)) &
    (master1["Column"] == "Number") &
    (master1["Entity Group"].isin(["Banks", "Non-Banks"]))
].copy()

arrears_90_plus = (
    arrears_90_plus
    .groupby("Reporting Date")["Amt"]
    .sum()
    .reset_index()
    .rename(columns={
        "Reporting Date": "date",
        "Amt": "pdh_arrears_90_plus_accounts"
    })
)


# ---------------------------------------------------------
# 4. Combine the three metrics
# ---------------------------------------------------------

mortgage_arrears = (
    outstanding
    .merge(total_arrears, on="date", how="outer")
    .merge(arrears_90_plus, on="date", how="outer")
    .sort_values("date")
    .reset_index(drop=True)
)


# ---------------------------------------------------------
# 5. Convert account counts to integers
# ---------------------------------------------------------

count_columns = [
    "pdh_mortgage_accounts",
    "pdh_arrears_accounts",
    "pdh_arrears_90_plus_accounts"
]

for column in count_columns:
    mortgage_arrears[column] = pd.to_numeric(
        mortgage_arrears[column],
        errors="coerce"
    )


# ---------------------------------------------------------
# 6. Derived arrears percentage
# ---------------------------------------------------------

mortgage_arrears["pdh_arrears_rate"] = (
    mortgage_arrears["pdh_arrears_accounts"]
    / mortgage_arrears["pdh_mortgage_accounts"]
    * 100
)


# ---------------------------------------------------------
# 7. Derived >90-day arrears percentage
# ---------------------------------------------------------

mortgage_arrears["pdh_arrears_90_plus_rate"] = (
    mortgage_arrears["pdh_arrears_90_plus_accounts"]
    / mortgage_arrears["pdh_mortgage_accounts"]
    * 100
)


# ---------------------------------------------------------
# 8. Final column order
# ---------------------------------------------------------

mortgage_arrears = mortgage_arrears[
    [
        "date",
        "pdh_mortgage_accounts",
        "pdh_arrears_accounts",
        "pdh_arrears_90_plus_accounts",
        "pdh_arrears_rate",
        "pdh_arrears_90_plus_rate"
    ]
]


# ---------------------------------------------------------
# 9. Save processed dataset
# ---------------------------------------------------------

output_path = "../data/processed/mortgage_arrears.csv"

mortgage_arrears.to_csv(
    output_path,
    index=False
)

print("Processed mortgage arrears data saved to:", output_path)
print("Shape:", mortgage_arrears.shape)
print()
print(mortgage_arrears.head())
print()
print(mortgage_arrears.tail())
```
