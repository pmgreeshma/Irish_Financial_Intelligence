```python
import pandas as pd

df = pd.read_csv(
    "../data/raw/hpm09_raw.csv",
    encoding="utf-8-sig"
)

df.columns = df.columns.str.replace(
    'ï»¿"',
    '',
    regex=False
)

df.columns = df.columns.str.replace(
    '"',
    '',
    regex=False
)

print("Rows:", len(df))
print("Columns:", len(df.columns))
print("\nColumn names:")
print(df.columns.tolist())
print(df.shape)
print(df.head())
```

```python
property_prices = df[
    (df["STATISTIC"] == "HPM09C01") &
    (df["Type of Residential Property"] == "National - all residential properties")
].copy()
```

```python
property_prices = df[
    (df["STATISTIC"] == "HPM09C01") &
    (df["Type of Residential Property"] == "National - all residential properties")
].copy()

print(property_prices.shape)
print(property_prices.head())
print(property_prices.tail())
```

```python
print("\nMissing values:")
print(property_prices[["Month", "VALUE"]].isna().sum())

print("\nDuplicate months:")
print(property_prices["Month"].duplicated().sum())
```

# create the processed dataset

```python
property_prices = property_prices[["Month", "VALUE"]].copy()

property_prices = property_prices.rename(columns={
    "Month": "date",
    "VALUE": "property_price_index"
})

property_prices["date"] = pd.to_datetime(
    property_prices["date"],
    format="%Y %B"
)

property_prices = property_prices.sort_values("date").reset_index(drop=True)

print(property_prices.head())
print(property_prices.tail())
print("Shape:", property_prices.shape)
```

# final validation

```python
print("Missing values:")
print(property_prices.isna().sum())

print("\nDuplicate dates:")
print(property_prices["date"].duplicated().sum())

print("\nDate range:")
print(property_prices["date"].min(), "to", property_prices["date"].max())

print("\nProperty price index range:")
print(
    property_prices["property_price_index"].min(),
    "to",
    property_prices["property_price_index"].max()
)
```

```python
property_prices.to_csv(
    "../data/processed/property_prices.csv",
    index=False
)

print("Saved to data/processed/property_prices.csv")
```
